## 1.1 Notation and basic definitions

### Definition 1.1.1 (Multiindex)

> The vector $\alpha = (\alpha_1,\dots,\alpha_n) \in \mathbb{N}_0^n$ is called a **multiindex** of order
>
> $$|\alpha| := \sum_{j=1}^{n} \alpha_j.$$
>
> If $x \in \mathbb{K}^n$, then we write $x^\alpha$ for the product $x_1^{\alpha_1} x_2^{\alpha_2}\cdots x_n^{\alpha_n} \in \mathbb{K}$.  
> For a function $u : \Omega \to \mathbb{K}$ and an arbitrary multiindex $\alpha$, the **$\alpha$-th partial derivative** is
>
> $$D^\alpha u \;:=\; \frac{\partial^{|\alpha|}}{\partial x_1^{\alpha_1}\cdots \partial x_n^{\alpha_n}}\,u \;=\; \underbrace{u_{x_1\cdots x_1}}_{\alpha_1\text{ times}}\underbrace{{}_{\cdots x_n\cdots x_n}}_{\alpha_n\text{ times}}.$$
>
> If $\alpha = 0$, then $D^0 u := u$.
>
> For $k \in \mathbb{N}_0$, $D^k u$ is shorthand for the vector of **all** partial derivatives of $u$ of order $k$:
> $$D^1 u \equiv Du = (u_{x_1},\dots,u_{x_n}),$$
> $$D^2 u = (u_{x_1 x_1}, u_{x_1 x_2},\dots, u_{x_1 x_n},\dots, u_{x_n x_n}),$$
> etc. For $D^2 u$ the same notation is also used for the **Hessian matrix** — context tells you which.


#### Elaboration — what this notation is *for*

A single PDE can involve many different partial derivatives: $u_{xx}$, $u_{xy}$, $u_{yyz}$, and so on. Writing them all out becomes unreadable fast. A **multiindex** is a bookkeeping device: one tuple $\alpha$ that simultaneously records (i) *which* variables you're differentiating with respect to, and (ii) *how many times* each.

Think of $\alpha$ as a histogram. In dimension $n=3$, the multiindex $\alpha = (2, 0, 1)$ means: *differentiate twice in $x_1$, not at all in $x_2$, once in $x_3$*. The **order** $|\alpha| = 2+0+1 = 3$ is the total number of derivatives taken — i.e., this is a third-order derivative.

The three notations the definition attaches to a multiindex:

1. **Monomial $x^\alpha$** — the product $x_1^{\alpha_1}\cdots x_n^{\alpha_n}$. So $(2,0,1)$ on $x=(x_1,x_2,x_3)$ gives $x_1^2 x_3$. Useful for writing Taylor series compactly.
2. **Derivative $D^\alpha u$** — the mixed partial derivative with that "signature". $(2,0,1)$ gives $\dfrac{\partial^3 u}{\partial x_1^2 \,\partial x_3}$.
3. **Order $|\alpha|$** — total number of derivatives (or total degree of the monomial). Lets us say things like "all derivatives of order $\le m$" just as $\sum_{|\alpha|\le m}$, no matter what $n$ is.

The entire point: the innocent-looking sum $\displaystyle\sum_{|\alpha|\le m} a_\alpha(x)\, D^\alpha u(x)$ will appear two pages from now as the definition of a general linear PDE of order $m$. Without multiindex notation you'd need a different formula for every $(n,m)$ pair.

#### Example 1 — unpacking multiindices in 2D

Let $n=2$ so $x = (x_1, x_2)$, and take $u(x_1, x_2) = x_1^3 x_2^2$. Then:

| $\alpha$ | $\lvert\alpha\rvert$ | $x^\alpha$ | $D^\alpha u$ |
|---|---|---|---|
| $(0,0)$ | 0 | $1$ | $u = x_1^3 x_2^2$ |
| $(1,0)$ | 1 | $x_1$ | $u_{x_1} = 3 x_1^2 x_2^2$ |
| $(0,1)$ | 1 | $x_2$ | $u_{x_2} = 2 x_1^3 x_2$ |
| $(2,0)$ | 2 | $x_1^2$ | $u_{x_1 x_1} = 6 x_1 x_2^2$ |
| $(1,1)$ | 2 | $x_1 x_2$ | $u_{x_1 x_2} = 6 x_1^2 x_2$ |
| $(0,2)$ | 2 | $x_2^2$ | $u_{x_2 x_2} = 2 x_1^3$ |
| $(3,2)$ | 5 | $x_1^3 x_2^2$ | $u_{x_1 x_1 x_1 x_2 x_2} = 12$ |

The last row is a nice sanity check: differentiating $x_1^3 x_2^2$ all the way down (three times in $x_1$, twice in $x_2$) gives $3! \cdot 2! = 12$, a constant, and any higher derivative is zero.

#### Example 2 — $D^k u$ as "the vector of all order-$k$ derivatives"

With $n = 3$ and $u(x_1, x_2, x_3)$ smooth:

- $D^1 u = D u = (u_{x_1},\,u_{x_2},\,u_{x_3})$ — this is just the gradient $\nabla u$.
- $D^2 u$ lists every second-order partial: $(u_{x_1 x_1},\,u_{x_1 x_2},\,u_{x_1 x_3},\,u_{x_2 x_2},\,u_{x_2 x_3},\,u_{x_3 x_3})$. Under the assumption of smoothness, mixed partials commute (Clairaut/Schwarz), so we only list each unordered pair once — 6 components total.
- As a **Hessian matrix** the same information is arranged as a symmetric $3\times 3$ matrix
  $$H_u = \begin{pmatrix} u_{x_1 x_1} & u_{x_1 x_2} & u_{x_1 x_3} \\ u_{x_1 x_2} & u_{x_2 x_2} & u_{x_2 x_3} \\ u_{x_1 x_3} & u_{x_2 x_3} & u_{x_3 x_3} \end{pmatrix}.$$

Kennedy's footnote is a warning: "$D^2 u$" is overloaded — sometimes a flat list, sometimes a matrix. The number of distinct order-$k$ partials of a function on $\mathbb{R}^n$ is $\binom{n+k-1}{k}$.

#### Example 3 — why this makes life bearable: Taylor's theorem

For a smooth $u:\mathbb{R}^n \to \mathbb{R}$, the Taylor expansion around $x_0$ to order $m$ in multiindex notation is simply

$$u(x_0 + h) \;=\; \sum_{|\alpha|\le m} \frac{D^\alpha u(x_0)}{\alpha!}\, h^\alpha \;+\; R_m(h),$$

where $\alpha! := \alpha_1!\,\alpha_2!\cdots \alpha_n!$. That single line replaces the ever-lengthening nested sums of the usual multi-variable Taylor formula. To appreciate the compression, try writing the degree-3 expansion on $\mathbb{R}^3$ by hand without multiindices — you'll produce roughly 20 distinct terms.

### Definition 1.1.2 (Partial differential equation)

> A **partial differential equation** (of order $m \in \mathbb{N}_0$), abbreviated **PDE**, is a relation of the form
>
> $$F\bigl(x,\,u(x),\,Du(x),\,\dots,\,D^m u(x)\bigr) = f(x), \qquad x \in \Omega, \tag{1.1.1}$$
>
> where $\Omega \subset \mathbb{R}^n$ is a given, usually open, set, $F : \Omega \times \mathbb{R} \times \mathbb{R}^n \times \cdots \times \mathbb{R}^{n^m} \to \mathbb{K}$ and $f : \Omega \to \mathbb{K}$ are given functions, and $u : \Omega \to \mathbb{K}$ is the unknown.
>
> Typically we impose auxiliary conditions on $u$ — either **boundary conditions** on $\partial\Omega$, or **conditions at infinity** (if $\Omega = \mathbb{R}^n$ or $\mathbb{R}^n_+$). We use the term *PDE* both for (1.1.1) alone and for (1.1.1) together with its auxiliary conditions.
>
> An expression of the form $F(x, u, \dots, D^m u)$ is called a **(partial) differential expression**, or loosely a **(partial) differential operator (PDO)**. We often denote it
>
> $$Lu := F(x, u, \dots, D^m u).$$
>
> Several PDEs in one or more unknowns constitute a **system**.

#### Elaboration — $F$ as a "slot-filling function"

The way to read $(1.1.1)$ is: $F$ is a fixed, garden-variety function of *many scalar arguments* — it doesn't know anything about differentiation. We feed it slot-by-slot:

- slot 1: the point $x \in \Omega$ (that's $n$ real numbers);
- slot 2: the value $u(x)$ (one real number);
- slot 3: all first derivatives $Du(x) = (u_{x_1},\dots,u_{x_n})$ ($n$ real numbers);
- slot 4: all second derivatives $D^2 u(x)$ ($n^2$ numbers if we list with repetition, or $\binom{n+1}{2}$ if we de-duplicate using symmetry);
- $\dots$
- slot $m+2$: all $m$-th derivatives $D^m u(x)$.

That's why the domain of $F$ is written as the cartesian product $\Omega \times \mathbb{R} \times \mathbb{R}^n \times \cdots \times \mathbb{R}^{n^m}$. The dimensions $1, n, n^2, \dots, n^m$ are just "how many slots does each derivative order need?" (The factor $n^m$ is an overcount that ignores the symmetry of mixed partials — Kennedy keeps it simple.)

**The order $m$ is the highest $k$ for which $F$ genuinely depends on $D^k u$.** If you could rewrite $F$ so that $D^m u$ never appears, the PDE would really be of order $< m$.

**Why introduce $L$?** Once you fix $F$, the map $u \mapsto F(x, u, Du, \dots, D^m u)$ is an operator that *eats a function and spits out another function*. Calling it $L$ (or $Lu$ for its value at $u$) lets us write the PDE simply as $Lu = f$, which hides the plumbing and makes PDE theory look like linear algebra: given an operator $L$ and a right-hand side $f$, solve $Lu = f$. Much of the course is about making this analogy rigorous.

**Important subtlety.** $F$ is just a function of numbers; $L$ is an operator on functions. The same $F$ gives the same $L$, but they live at different levels of abstraction. Getting comfortable switching between them is half the battle.

#### Example — Poisson's equation fits the mould

Take $n = 2$, $\Omega \subset \mathbb{R}^2$, and Poisson's equation $-\Delta u = f$, i.e.

$$-u_{x_1 x_1}(x) - u_{x_2 x_2}(x) = f(x).$$

To fit this into $(1.1.1)$ we need to identify $F$. Slots for order $m=2$ and $n=2$ are: $x \in \mathbb{R}^2$, $u \in \mathbb{R}$, $Du \in \mathbb{R}^2$, $D^2 u \in \mathbb{R}^4$ (listing $u_{x_1 x_1}, u_{x_1 x_2}, u_{x_2 x_1}, u_{x_2 x_2}$ with repetition).

Introduce dummy names for those slot values: call them $\xi \in \mathbb{R}^2$ (the would-be gradient) and $\eta = (\eta_{11}, \eta_{12}, \eta_{21}, \eta_{22}) \in \mathbb{R}^4$ (the would-be Hessian entries). Then define

$$F(x, z, \xi, \eta) \;:=\; -\eta_{11} - \eta_{22}.$$

That's it. $F$ is literally the function that adds two of its $\eta$-slot entries and flips the sign. It doesn't depend on $x$, on $z$ (the $u$-slot), or on $\xi$ (the gradient slot). Substituting the actual derivatives of $u$ back in,

$$F(x, u(x), Du(x), D^2 u(x)) = -u_{x_1 x_1}(x) - u_{x_2 x_2}(x) = -\Delta u(x),$$

and setting this equal to $f(x)$ recovers Poisson's equation. The associated operator is $L u := -\Delta u$.

Notice: $F$ is linear in $\eta$ and independent of $\xi$ and $z$. Two pages from now that property will be the definition of a *linear* PDE.

#### Python realization

Let's make the abstract machinery concrete with SymPy. We'll (1) build a symbolic $u$, (2) compute $D^\alpha u$ for arbitrary multiindices, (3) build the Poisson operator $L u = -\Delta u$, (4) verify it's equivalent to a function $F$ of the slot values, and (5) write a helper that reports the order $m$ of an operator by looking at the highest derivative it contains.

```python
import sympy as sp

# Set up coordinates and an abstract smooth function u(x1, x2).
x1, x2 = sp.symbols('x1 x2', real=True)
u = sp.Function('u')(x1, x2)

def D_alpha(expr, variables, alpha):
    """Apply the multiindex derivative D^alpha. variables is a tuple (x1,...,xn)."""
    assert len(variables) == len(alpha)
    out = expr
    for var, k in zip(variables, alpha):
        if k:
            out = sp.diff(out, var, k)
    return out

# Sanity check against Example 1 from Definition 1.1.1
poly = x1**3 * x2**2
for alpha in [(0,0), (1,0), (0,1), (2,0), (1,1), (0,2), (3,2)]:
    print(f"alpha={alpha}  |alpha|={sum(alpha)}  D^alpha(x1^3 x2^2) = {D_alpha(poly, (x1,x2), alpha)}")
```

```python
# Build the Poisson operator L u = -Delta u as an operator on functions.
def L_poisson(u_expr, variables):
    return -sum(sp.diff(u_expr, v, 2) for v in variables)

# Test on a specific smooth u.
u_concrete = sp.sin(x1) * sp.cos(x2)
print("L[sin(x1)cos(x2)] =", sp.simplify(L_poisson(u_concrete, (x1, x2))))
# Expected: 2 sin(x1) cos(x2) -- recovering the eigenfunction relation from Q6 of 1.1.1.
```

```python
# Now demonstrate F as a plain numeric function of slot values.
# Slots for n=2, m=2: x=(x1,x2), z=u, xi=(u_x1,u_x2), eta=(u_x1x1,u_x1x2,u_x2x1,u_x2x2).
def F_poisson(x, z, xi, eta):
    # F depends only on eta[0] (=u_x1x1) and eta[3] (=u_x2x2).
    return -eta[0] - eta[3]

# Feed it the derivatives of u_concrete symbolically and confirm it matches L u.
slot_x  = (x1, x2)
slot_z  = u_concrete
slot_xi = (sp.diff(u_concrete, x1), sp.diff(u_concrete, x2))
slot_eta = (sp.diff(u_concrete, x1, 2),
            sp.diff(u_concrete, x1, x2),
            sp.diff(u_concrete, x2, x1),
            sp.diff(u_concrete, x2, 2))

via_F = F_poisson(slot_x, slot_z, slot_xi, slot_eta)
via_L = L_poisson(u_concrete, (x1, x2))
print("via F:", sp.simplify(via_F))
print("via L:", sp.simplify(via_L))
print("equal?", sp.simplify(via_F - via_L) == 0)
```

```python
# A helper that reports the order m of an operator by probing it with
# a generic symbolic u and inspecting which derivatives appear.
def operator_order(L_callable, variables):
    """Apply L to an abstract u and return the highest derivative order present."""
    u_abs = sp.Function('u')(*variables)
    expr = L_callable(u_abs, variables)
    max_order = 0
    for deriv in expr.atoms(sp.Derivative):
        order = sum(count for _, count in deriv.variable_count)
        max_order = max(max_order, order)
    return max_order

print("order of -Delta:", operator_order(L_poisson, (x1, x2)))

# Biharmonic operator L u = Delta^2 u -- should report order 4.
def L_biharmonic(u_expr, variables):
    lap = sum(sp.diff(u_expr, v, 2) for v in variables)
    return sum(sp.diff(lap, v, 2) for v in variables)

print("order of Delta^2:", operator_order(L_biharmonic, (x1, x2)))

# A first-order transport operator L u = u_x1 + u_x2.
def L_transport(u_expr, variables):
    return sum(sp.diff(u_expr, v) for v in variables)

print("order of transport:", operator_order(L_transport, (x1, x2)))
```

```python
# Answers
#
# 1. Heat equation, n=2 with coordinates (t, x1). Order m=2.
#    Slots: point=(t, x1), z=u, xi=(u_t, u_x1), eta=(u_tt, u_t x1, u_x1 t, u_x1 x1).
#    F(point, z, xi, eta) = xi[0] - eta[3]     (i.e. u_t - u_x1 x1)
#    Depends on: xi[0] (=u_t) and eta[3] (=u_x1 x1). Independent of point, z, xi[1], eta[0..2].
#    Note u_t is a FIRST derivative, so it sits in the xi-slot even though the equation
#    is overall 'second order'. The order m is determined by the highest slot used.
#
# 2. Burgers, n=2 with coordinates (t, x1). Order m=1.
#    Slots: point=(t, x1), z=u, xi=(u_t, u_x1).
#    F(point, z, xi) = xi[0] + z * xi[1]     (i.e. u_t + u * u_x1)
#    Nonlinearity: F contains the product z * xi[1]. It depends on z (the u-slot) AND on
#    xi[1], and they multiply -- not linear in (z, xi) jointly. That product is what makes
#    Burgers famously hard: shocks form in finite time.
#
# 3. Literally-true-but-spirit-false. u_{x1 x2} - u_{x2 x1} vanishes identically for smooth u
#    (Schwarz's theorem), so the operator is really the zero operator of order 0. The
#    definition's phrase 'depends on D^m u' should be read as 'depends non-trivially'.
#    Always ask whether the apparent highest-order term actually does something.
#
# 4. Delta^2 u in n=2:
#    Delta u = u_x1x1 + u_x2x2
#    Delta^2 u = (u_x1x1 + u_x2x2)_x1x1 + (u_x1x1 + u_x2x2)_x2x2
#              = u_x1x1x1x1 + u_x1x1x2x2 + u_x1x1x2x2 + u_x2x2x2x2
#              = u_x1x1x1x1 + 2 u_x1x1x2x2 + u_x2x2x2x2.
#    Three distinct fourth-order terms with coefficients 1, 2, 1.
#
# 5. Order m = 2. F is linear in eta (for fixed xi) and linear in xi (for fixed eta), but
#    not jointly -- products xi_i * eta_j appear. Crucially, F IS linear in the
#    HIGHEST-order slot eta alone, with coefficients that depend on lower-order data (xi).
#    That is the definition of quasilinear.

# Q6: enhanced operator_order
import sympy as sp
from collections import Counter
x1, x2 = sp.symbols('x1 x2', real=True)

def operator_profile(L_callable, variables):
    u_abs = sp.Function('u')(*variables)
    expr = sp.expand(L_callable(u_abs, variables))
    multiindices_seen = Counter()
    for deriv in expr.atoms(sp.Derivative):
        counts = dict(deriv.variable_count)
        alpha = tuple(counts.get(v, 0) for v in variables)
        multiindices_seen[alpha] += 1
    order = max((sum(a) for a in multiindices_seen), default=0)
    return order, sorted(multiindices_seen)

def L_biharmonic(u_expr, variables):
    lap = sum(sp.diff(u_expr, v, 2) for v in variables)
    return sum(sp.diff(lap, v, 2) for v in variables)

order, alphas = operator_profile(L_biharmonic, (x1, x2))
print(f"Biharmonic: order={order}, multiindices={alphas}")
# Expect order 4 and multiindices (4,0), (2,2), (0,4) -- matching Q4's three distinct terms.
```
